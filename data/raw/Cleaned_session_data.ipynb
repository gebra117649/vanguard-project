```python
# To start analyzing the data, we want to create a table with a row for each session.

import pandas as pd

web_data = pd.read_csv("cleaned_web_data.csv")
```

```python
web_data.head(10)
```

```python
web_data = web_data.drop(columns="Unnamed: 0")
```

```python
web_data.info()
```

```python
sessions = (
    web_data.groupby(["client_id", "visit_id"])
    .size()
    .reset_index(name="step_count")
)

sessions.head(10)
```

```python
sessions.to_csv("sessions_data.csv", sep=",", encoding="utf-8", index=False)
```
