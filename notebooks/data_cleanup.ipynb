```python
import pandas as pd
```

```python
web_data_1 = pd.read_csv("df_final_web_data_pt_1.txt")
```

```python
web_data_2 = pd.read_csv("df_final_web_data_pt_2.txt")
```

```python
experiment_clients = pd.read_csv("df_final_experiment_clients.txt")
```

```python
final_demo = pd.read_csv("df_final_demo.txt")
```

```python
web_data_1.info()
```

```python
web_data_2.info()
experiment_clients.info()
final_demo.info()
```

```python
experiment_clients.head(10)
```

```python
web_data = pd.concat([web_data_1, web_data_2], ignore_index=True)

web_data.head(10)
```

```python
merged_web_data = pd.merge(web_data, experiment_clients, on="client_id", how="outer")
merged_web_data.head(10)
```

```python
merged_web_data.info()
```

```python
client_demo = pd.merge(final_demo, experiment_clients, on="client_id", how="outer")
client_demo.head(10)
client_demo.info()
```

```python
# Check for duplicated records 
merged_web_data.duplicated().sum()
```

```python
client_demo.duplicated().sum()
```

```python
# Remove duplicated records

merged_web_data = merged_web_data.drop_duplicates().reset_index(drop=True)
```

```python
merged_web_data.duplicated().sum()
```

```python
# Check for empty values
merged_web_data.isna().sum()
```

```python
client_demo.isna().sum()
```

```python
# We drop records with no value for Variation as they're not taking part in the experiment

merged_web_data = merged_web_data.dropna()

# We removed all records with empty values, as only the Variation column has empty values, so no need to specify.
```

```python
merged_web_data.info()
```

```python
# We drop records with empty values for variation

client_demo = client_demo.dropna(subset=["Variation"])
```

```python
client_demo.info()
```

```python
client_demo.isna().head(15)
```

```python
# We drop all records with empty client_id 
client_demo = client_demo.dropna(subset=["client_id"])
```

```python
client_demo.info()
```

```python
# We maintain users with empty values related to demo information as this isn't relevant to the main analysis and the number is very low.
# We name the empty values in gender as Unknown

client_demo["gendr"] = client_demo["gendr"].fillna("U")
```

```python
client_demo.info()
```

```python
client_demo["gendr"].unique()
```

```python
client_demo["gendr"] = client_demo["gendr"].replace({
    "X": "U"
    })
```

```python
client_demo["gendr"].unique()
```

```python
# We standardize and clean column names
client_demo.columns = (client_demo.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","")
    .str.replace("gendr", "gender")
    .str.replace("clnt_tenure_yr", "tenure_year")
    .str.replace("clnt_tenure_mnth", "tenure_month")
    .str.replace("clnt_age", "age")
    .str.replace("num_accts", "accounts")
    .str.replace("calls_6_mnth", "calls")
    .str.replace("logons_6_mnth", "logons")
    .str.replace("bal", "balance")
)
client_demo.info()
```

```python
merged_web_data.columns = (merged_web_data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","")
)
merged_web_data.info()
```

```python
# We export the datasets to analyze them further

merged_web_data.to_csv("cleaned_web_data.csv", sep=",", encoding="utf-8")
client_demo.to_csv("cleaned_client_data.csv", sep=",", encoding="utf-8")
```
