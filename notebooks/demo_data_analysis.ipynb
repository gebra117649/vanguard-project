```python
import pandas as pd

client_demo = pd.read_csv("cleaned_client_data.csv")
```

```python
client_demo.head(10)
```

```python
# Drop Unnamed column

client_demo = client_demo.drop(columns="Unnamed: 0")
```

```python
client_demo.info()
```

```python
# Understand the percentage of test vs control users

group_sizes = client_demo["variation"].value_counts()

display(group_sizes.to_frame("clients").assign(
    distribution_percentage = (100 * group_sizes / group_sizes.sum()).round(2)
))
```

```python
# We want to understand the demographics of the users participating in the experiment
numeric_columns = ["age", "balance", "tenure_year", "accounts", "calls", "logons"]

for column in numeric_columns:
    print(column)
    display(client_demo.groupby("variation")[column].describe().round(2))
```

```python
# Checking the distribution of clients by gender
gender_distribution = (pd.crosstab(
    client_demo["gender"],
    client_demo["variation"],
    normalize="columns",
)*100).round(2)

display(gender_distribution)
```
